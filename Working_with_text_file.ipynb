```python
# 1. Creating and Writing to a Text File

# Open file in write mode

file = open("students.txt", "w")

file.write("Rahul\n")
file.write("Priya\n")
file.write("Amit\n")
file.write("Neha\n")

file.close()

print("File created seccessfully")
```

```python
# 2. Writing Multiple Lines Using writelines()

students = [
    "Rahul\n",
    "Priya\n",
    "Amit\n",
    "Neha\n"
]

with open ("students.txt", "w") as file:
  file.writelines(students)

print("Data written successfully")
```

```python
# 3. Reading the entire file.
with open("students.txt","r") as file:
  data=file.read()

print(data)
```

```python
# 4.reding file line by line

with open("students.txt","r") as file:
  for line in file:
    print(line.strip()) #print(line)
```

strip() remove the newline character(/n) from each line.

```python
# 5. using readline()

with open ("students.txt","r") as file:
  line1 =file.readline()
  line2=file.readline()

print(line1.strip())
print(line2.strip())
```

```python
# 6. using readline()

with open ("students.txt","r") as file:
  lines =file.readlines()
print(lines)
```

```python
# you can also process the list:
with open("students.txt","r") as file:
  lines = file.readlines()
for line in lines:
  print(line.strip())
```

```python
# 7. Appending data to a file

with open("students.txt","a") as file:
  file.write("Karan\n")
  file.write("Simran\n")

print("New students added")
```

```python
# 8. Writing user input to a file.
name = input("Enter a name:")
age = input("Enter a age:")

with open("user.txt","w") as file:
  file.write("Name:"+name+"\n")
  file.write("Age:"+age+"\n")

print("information saved.")
```

```python
# 9. counting the no lines.
with open ("students.txt","r") as file:
  lines=file.readlines()

print("Total no of lines:",len(lines))
```

```python
# 10. counting words in a text file .
with open("students.txt","r") as file:
  data=file.read()

words=data.split()
print("Total no of words:",len(words))
```

```python
# 11. counting characters

with open ("students.txt","r") as file:
  data=file.read()

print("Total no of characters:",len(data))
```

```python
# to exclude space:
characters = data.replace("","").replace(" ","")
print("Total no of characters:",len(characters))
```

```python
# 12. searching for a word ina text file.
search_word = input("Enter a word to search:")

with open ("students.txt","r") as file:
  data=file.read()

if search_word in data:
  print("Found")
else:
  print("Word not found")
```

```python
# 13. counting occurence of a word
search_word = input("Enter a word to search:")

with open ("students.txt","r") as file:
  data=file.read()

count=data.lower().count(search_word.lower())
print("Occurence:",count)
```

```python
# 14. copying content from one file to another

with open("students.txt","r") as source:
  data = source.read()

with open("backup.txt","w") as destination:
  destination.write(data)

print("File copied successfully")
```
